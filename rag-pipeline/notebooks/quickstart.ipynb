# 🔍 RAG Pipeline — Quickstart Notebook

This notebook walks through the full RAG pipeline:
1. **Load** sample text
2. **Ingest** (chunk → embed → index)
3. **Retrieve** relevant chunks
4. **Generate** a grounded answer

```python
from src.rag_pipeline import RAGPipeline

rag = RAGPipeline()
```

```python
# Write a quick sample document to test with
sample = """Retrieval-Augmented Generation (RAG) is a technique that combines
dense retrieval with large language model generation. Instead of relying solely
on parametric knowledge baked into the LLM weights, RAG retrieves relevant
documents at inference time and conditions the generation on them.

FAISS (Facebook AI Similarity Search) is an efficient library for similarity
search and clustering of dense vectors. It supports exact and approximate
nearest-neighbour search and scales to billions of vectors.

Sentence Transformers are transformer models fine-tuned to produce semantically
meaningful sentence embeddings. The all-MiniLM-L6-v2 model maps sentences to a
384-dimensional dense vector space."""

with open('../data/sample.txt', 'w') as f:
    f.write(sample)

rag.ingest('../data/sample.txt')
```

```python
# Retrieve without generating — useful for debugging
chunks = rag.retrieve('What is FAISS?', top_k=2)
for c in chunks:
    print(f"Score: {c['score']:.3f}")
    print(c['text'])
    print()
```

```python
# Full RAG query
result = rag.query('Explain RAG in simple terms.', verbose=True)
print('\nAnswer:', result['answer'])
print('Sources:', result['sources'])
```

```python
# Save index for reuse
rag.save_index('../index')
```
